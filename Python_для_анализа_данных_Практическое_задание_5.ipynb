# **Polars**

```python
import polars as pl
```

```python
# Загрузка датасета с помощью polars
df = pl.read_csv("train.csv")
```

```python
# Выводим информацию о типах данных и количестве пропусков
print("Типы данных:")
print(df.dtypes)

print("\nКоличество пропусков:")
print(df.null_count())

print("\nОсновные статистики:")
print(df.describe())

print("\nСредние значения числовых столбцов:")
print(df.mean())
```

```python
# Просчет количества пассажиров каждого класса (Pclass)
pclass_counts = df.get_column('Pclass').value_counts().sort(by='Pclass')
pclass_counts
```

```python
# Количество выживших мужчин и женщин на корабле
survival_counts = df.group_by("Sex").agg(
    pl.col("Survived").sum())
survival_counts
```

```python
# Пассажиры старше 44 лет
df.filter(pl.col("Age") > 44)
```

# **Ускорение работы с pandas**

```python
import pandas as pd
import bottleneck as bn
```

```python
# Загрузка датасета
df = pd.read_csv("train.csv")
```

```python
# Подсчет среднего возраста пассажиров и его стандартного отклонения с использованием bottleneck
mean_age = bn.nanmean(df["Age"])
std_age = bn.nanstd(df["Age"])
print(f"Средний возраст пассажиров: {mean_age}")
print(f"Стандартное отклонение возраста: {std_age}")
```

```python
# Умножение значений столбца Fare на 1.3 и сохранение результатов в Fare_new
def multiply_by_1_3(x):
    return x * 1.3

df["Fare_new"] = df["Fare"].apply(multiply_by_1_3)
df["Fare_new"]
```

# **Оптимизация типов pandas**

```python
# Загружаем датасет
df = pd.read_csv('Housing.csv')
```

```python
# Сохраняем исходное потребление памяти
memory_before = df.memory_usage(deep=True).sum() / (1024 ** 2)  # в МБ
print(f"Потребление памяти до оптимизации: {memory_before:.2f} МБ")
```

```python
# Получение информации об исходном датасете: пропусках и типах данных
df.info()
```

```python
# Анализ используемых значений
for col in df.columns:
    print(f"Столбец {col}: уникальных значений - {df[col].nunique()}, min - {df[col].min()}, max - {df[col].max()}, пример значений - {df[col].unique()[:5]}")
```

Столбец price - исходный тип данных int64. Данный столбец содержит положительные числовые значения в диапазоне от 1750000 до 13300000. Описанный диапазон значений помещается в int32.
Столбец area - исходный тип данных int64. Данный столбец содержит положительные числовые значения в диапазоне от 1650 до 16200. Описанный диапазон значений помещается в int16.
Столбец bedrooms - исходный тип данных int64. Данный столбец содержит положительные числовые значения в диапазоне от 1 до 6. Описанный диапазон значений помещается в int8.
Столбец bathrooms - исходный тип данных int64. Данный столбец содержит положительные числовые значения в диапазоне от 1 до 4. Описанный диапазон значений помещается в int8.
Столбец stories - исходный тип данных int64. Данный столбец содержит положительные числовые значения в диапазоне от 1 до 4. Описанный диапазон значений помещается в int8.
Столбец parking - исходный тип данных int64. Данный столбец содержит неотрицательные числовые значения в диапазоне от 0 до 3. Описанный диапазон значений помещается в int8.
Столбцы mainroad, guestroom, basement, hotwaterheating, airconditioning, prefarea  - исходный тип данных object. Данные столбцы содержат строки 'yes', 'no'. Это категориальные признаки, которые могут быть представлены булевой переменной: 'yes' - True, 'no' - False.
Столбец furnishingstatus - исходный тип данных object. Данный стобец содержит строки 'furnished', 'semi-furnished', 'unfurnished'. Это категориальные признаки, которые могут быть представлены типом данных category.



```python
binary_cols = ['mainroad', 'guestroom', 'basement', 'hotwaterheating', 'airconditioning', 'prefarea']
for col in binary_cols:
    df[col] = df[col].replace({'yes': True, 'no': False}).astype('bool')
```

```python
df['price'] = df['price'].astype('int32')
df['area'] = df['area'].astype('int16')
df['bedrooms'] = df['bedrooms'].astype('int8')
df['bathrooms'] = df['bathrooms'].astype('int8')
df['stories'] = df['stories'].astype('int8')
df['parking'] = df['parking'].astype('int8')
```

```python
df['furnishingstatus'] = df['furnishingstatus'].astype('category')
```

```python
# Получение информации о преобразованном датасете: пропусках и типах данных
df.info()
```

```python
memory_after = df.memory_usage(deep=True).sum() / (1024 ** 2)  # в МБ
print(f"Потребление памяти после оптимизации: {memory_after:.2f} МБ")
```

```python
memory_savings = (memory_before - memory_after) / memory_before * 100
print(f"Экономия памяти: {memory_savings:.2f}%")
```
