```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import f1_score
from sklearn.dummy import DummyClassifier
from sklearn.preprocessing import StandardScaler
```

```python
RANDOM_STATE = 2025
```

```python
# Загрузка данных
data = pd.read_csv('train.csv')
```

Для предсказания были выбраны признаки 'Pclass', 'Sex', 'Age', 'SibSp', 'Parch', 'Fare' на основе следующего:
*   Pclass (Класс билета): Логично предположить, что класс, в котором ехал пассажир, мог влиять на его шансы на выживание: у пассажиров первого класса, вероятно, был лучший доступ к спасательным шлюпкам.
*   Sex (Пол): Женщинам и детям отдавался приоритет при спасении.
*   Age (Возраст): Как и пол, возраст также мог влиять на приоритет при спасении.
*   SibSp (Количество братьев/сестер/супругов на борту): Наличие членов семьи на борту могло повлиять на поведение пассажира и, следовательно, на его шансы на выживание (например, мог попытаться найти своих родственников вместо того, чтобы сразу же попасть в спасательную шлюпку).
*   Parch (Количество родителей/детей на борту): Аналогично SibSp.
*   Fare (Стоимость билета): Более дорогая стоимость билета, вероятно, указывала на более высокий класс.










```python
# Выбор признаков для предсказания
features = ['Pclass', 'Sex', 'Age', 'SibSp', 'Parch', 'Fare']
target = 'Survived'
```

```python
# Обработка пропущенных значений
data['Age'] = data['Age'].fillna(data['Age'].median())
data['Fare'] = data['Fare'].fillna(data['Fare'].median())
# Кодирование категориальных признаков
data['Sex'] = data['Sex'].map({'male': 0, 'female': 1})
```

```python
# Удаление строк с пропущенными значениями (на всякий случай)
data = data.dropna(subset=features)
```

```python
# Разделение на тренировочную и тестовую выборки
X_train, X_test, y_train, y_test = train_test_split(data[features], data[target], test_size=0.2, random_state=RANDOM_STATE)
```

```python
# Масштабирование признаков
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)  # fit_transform на тренировочных данных
X_test = scaler.transform(X_test)  # transform на тестовых данных
```

```python
# Создание бейзлайн модели
dummy_clf = DummyClassifier(strategy="most_frequent")
dummy_clf.fit(X_train, y_train)
dummy_predictions = dummy_clf.predict(X_test)
```

```python
# Создание и обучение модели
model = LogisticRegression(random_state=RANDOM_STATE)
model.fit(X_train, y_train)
# Предсказание на тестовой выборке
predictions = model.predict(X_test)
```

В данном случае метрикой для измерения данных была выбрана f1-score на основе следующего:
*   Существует некоторый дисбаланс классов: Количество погибших пассажиров немного больше, чем количество выживших.
*   precision важен также, как и recall.

```python
# Оценка качества бейзлайн модели
print("Baseline F1-score:", f1_score(y_test, dummy_predictions))
# Оценка качества модели
print("Model F1-score:", f1_score(y_test, predictions))
```
