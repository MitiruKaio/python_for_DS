```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
```

```python
# Базовый URL для веб-сайта Lifehacker.ru рубрики technology
base_url = "https://lifehacker.ru/topics/technology"
articles_data = []

# Перебираем первые 10 страниц рубрики
for page in range(1, 11):
    try:
        # Формируем URL для каждой страницы (определяем формат ссылки для пагинации)
        url = f"{base_url}/?page={page}"
        print(f"Обрабатывается страница: {url}")

        # Отправляем GET-запрос на текущую страницу
        response = requests.get(url)

        # Создаем объект BeautifulSoup для разбора HTML-контента
        soup = BeautifulSoup(response.text, 'lxml')

        # Находим все теги ссылок с классом 'lh-small-article-card__link'
        raw_items = soup.find_all('a', class_='lh-small-article-card__link')

        # Извлекаем заголовки статей
        titles = [item.get('title').replace("&nbsp;", ' ') for item in raw_items]

        # Извлекаем ссылки на статьи
        links = [item.get('href') for item in raw_items]

        # Cобираем содержимое статей
        for i, link in enumerate(links):
            try:

                full_link = "https://lifehacker.ru" + link
                response_article = requests.get(full_link)
                response_article.encoding = 'utf-8'
                soup_article = BeautifulSoup(response_article.text, 'lxml')

                # Находим все div объекты с классом 'single-article__post-content single-article__content-container'
                article = soup_article.find("div", class_="single-article__post-content single-article__content-container")

                if article:
                    # Извлекаем текст
                    article_text = article.text

                    articles_data.append({
                        'link': full_link,
                        'title': titles[i],
                        'text': article_text
                    })

            except requests.exceptions.RequestException as e:
                print(f"Ошибка при запросе статьи {full_link}: {e}")

    except requests.exceptions.RequestException as e:
        print(f"Ошибка при запросе страницы {url}: {e}")
```

```python
# Собираем полученные данные в DataFrame
df = pd.DataFrame(articles_data)
df, df.info(), df.head()
```
